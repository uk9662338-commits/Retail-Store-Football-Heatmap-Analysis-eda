```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Sample Footfall Data
data = {
    'X': [1,1,1,2,2,2,3,3,3,4,4],
    'Y': [1,2,3,1,2,3,1,2,3,2,3],
    'Footfall': [10,20,30,15,40,25,35,45,20,30,50]
}

df = pd.DataFrame(data)
df
```

```python
# heatmap format
heatmap_data = df.pivot(index='Y', columns='X', values='Footfall')
heatmap_data
```

```python
# Heatmap draw  (MAIN PART)
plt.figure(figsize=(6,5))
plt.imshow(heatmap_data, cmap='hot')
plt.colorbar(label='Customer Footfall')

plt.xlabel("Store Section (X)")
plt.ylabel("Store Section (Y)")
plt.title("Retail Store Footfall Heatmap")

plt.show()
```
